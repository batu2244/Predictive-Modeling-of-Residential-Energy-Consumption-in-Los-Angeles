```python
# Reading the 'api_locations' file that I have created using the API
import pandas as pd

api_locations = pd.read_csv('api_locations.csv', index_col=0)

# POST PROCESSING - important
# Making sure that the dictionaries in the dataframe are recognized by the python
import ast
# https://stackoverflow.com/questions/988228/convert-a-string-representation-of-a-dictionary-to-a-dictionary
# I have found this library from stackoverflow, which stated that 'as.literal_eval' as a safer alternative than just using 'eval' library.

print(isinstance(api_locations['raw_result'].iloc[0], dict))

# I need a double check for this because this gives an error when the code already is a dictionary.
if isinstance(api_locations['raw_result'].iloc[0], str):
    api_locations['raw_result'] = api_locations['raw_result'].apply(ast.literal_eval)

print(isinstance(api_locations['raw_result'].iloc[0], dict))
display(api_locations)
```

```python
# Creating a function that extracts the relevant information from the dictionary in 'raw result'
def extract_location_components(raw_result):
    try:
        first_result = raw_result['data']['results'][0] # Using the first capture
        components = first_result['address_components']
        geometry = first_result['geometry']['location']

        # Initialize output
        output = {
            "postal_code": None,
            "city": None,
            "county": None,
            "neighborhood": None,
            "lat": geometry.get('lat'),
            "lng": geometry.get('lng')
        }

        for comp in components:
            types = comp['types']
            if 'postal_code' in types:
                output['postal_code'] = comp['long_name']
            if 'locality' in types:
                output['city'] = comp['long_name']
            if 'administrative_area_level_2' in types:
                output['county'] = comp['long_name']
            if 'neighborhood' in types:
                output['neighborhood'] = comp['long_name']

        return output
    except:
        return {
            "postal_code": None,
            "city": None,
            "county": None,
            "neighborhood": None,
            "lat": None,
            "lng": None
        }
```

```python
# Applying this function to every row to generate the 'location details' column
api_locations['location_details'] = api_locations['raw_result'].apply(extract_location_components)
display(api_locations)
# I do have 271 neighborhood information
```
